```python
# 2 Matrix multiplication
# 2.1
import numpy as np
M1 = np.random.randint(0, 51, 50).reshape(5, 10)
M2 = np.random.randint(0, 51, 50).reshape(10, 5)
print(M1)
print(M2)
# 2.2
def Matrix_multip(M1, M2):
    r1, c1 = M1.shape
    r2, c2 = M2.shape
    p = np.zeros((r1, c2))
    if c1 != r2:
        print("M1列数必须与M2行数相等")
    else:
        for i in np.arange(r1):
            for j in np.arange(c2):
                for k in np.arange(c1):
                    p[i, j] = p[i, j] + M1[i, k] * M2[k, j]
    return p

print(Matrix_multip(M1, M2))
```
