```python
# 5
# 5.1
def Find_expression(x, p):   
    # p在此处起到控制输出的作用，主要为了5.2服务：
    # 如果p取1，则输出所有满足条件的解的表达式，如果p为其他值，则只输出一共有多少种解数n。
    sym = ["", "+", "-"]
    n = 0
    for i1 in sym:
        for i2 in sym:
            for i3 in sym:
                for i4 in sym:
                    for i5 in sym:
                        for i6 in sym:
                            for i7 in sym:
                                for i8 in sym:
                                    expr = f"1{i1}2{i2}3{i3}4{i4}5{i5}6{i6}7{i7}8{i8}9"
                                    if eval(expr) == x:
                                        if p == 1:
                                            print(f"{expr} = {x}")
                                        n = n + 1
    return n

print(Find_expression(50,1))
# 5.2
Total_solutions = [Find_expression(i+1, 0) for i in range(100)]

import matplotlib.pyplot as plt
plt.plot(Total_solutions)

maxnum = [i + 1 for i in range(100) if Total_solutions[i] == max(Total_solutions)]
minnum = [i + 1 for i in range(100) if Total_solutions[i] == min(Total_solutions)]
print(f"{maxnum} yields the maximum of Total_solutions")
print(f"{minnum} yields the minimum of Total_solutions")
```
